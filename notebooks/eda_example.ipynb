# Exploratory Data Analysis Example

This notebook demonstrates how to use the data analytics utilities in this repository for exploratory data analysis.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import custom modules
import sys
sys.path.append('src')

from preprocessing import handle_missing_values, remove_outliers_iqr, normalize_data
from eda import generate_summary, plot_distribution, plot_correlation_heatmap, missing_values_plot

# Set plotting style
sns.set_style('whitegrid')
%matplotlib inline
```

## Load Sample Data

For this example, we'll create a sample dataset. In practice, you would load your own data from CSV, Excel, or a database.

```python
# Create sample dataset
np.random.seed(42)
n_samples = 1000

data = {
    'age': np.random.normal(35, 12, n_samples),
    'income': np.random.normal(75000, 25000, n_samples),
    'spending_score': np.random.randint(1, 100, n_samples),
    'category': np.random.choice(['A', 'B', 'C', 'D'], n_samples),
    'years_experience': np.random.randint(0, 30, n_samples)
}

# Add some missing values
df = pd.DataFrame(data)
df.loc[np.random.choice(df.index, 50), 'income'] = np.nan
df.loc[np.random.choice(df.index, 30), 'age'] = np.nan

print(f"Dataset shape: {df.shape}")
df.head()
```

## Generate Summary Statistics

```python
# Generate summary
summary = generate_summary(df)

print("=== Dataset Overview ===")
print(f"Shape: {summary['shape']}")
print(f"\nColumns: {summary['columns']}")
print(f"\nData Types:\n{summary['dtypes']}")
print(f"\nMissing Values:\n{summary['missing_values']}")
print(f"\nMissing Percentage:\n{summary['missing_percentage']}")
```

## Visualize Missing Values

```python
# Plot missing values
fig, ax = missing_values_plot(df)
plt.show()
```

## Data Preprocessing

```python
# Handle missing values
df_clean = handle_missing_values(df, strategy='median')

print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# Remove outliers from income column
df_clean = remove_outliers_iqr(df_clean, 'income', multiplier=1.5)
print(f"Dataset shape after outlier removal: {df_clean.shape}")
```

## Visualize Distributions

```python
# Plot distribution of income
fig, ax = plot_distribution(df_clean, 'income', bins=40)
plt.show()
```

```python
# Plot distribution of age
fig, ax = plot_distribution(df_clean, 'age', bins=30)
plt.show()
```

## Correlation Analysis

```python
# Plot correlation heatmap
fig, ax = plot_correlation_heatmap(df_clean)
plt.show()
```

## Data Normalization

```python
# Normalize numerical columns
df_normalized = normalize_data(df_clean, method='minmax')

print("Normalized data summary:")
df_normalized.describe()
```
